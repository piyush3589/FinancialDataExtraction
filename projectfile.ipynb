```python
import pandas as pd
import re

df = pd.read_csv('financial_transactions_5000.csv')

# Extraction & cleaning
extracted = []
errors = []

for idx, row in df.iterrows():
    try:
        # Normalize merchant name
        merchant_clean = row['merchant'].strip().title()

        # Extract amount (validate)
        amount = float(row['amount'])
        if amount <= 0:
            errors.append({'id': row['transaction_id'], 'error': 'Invalid amount'})
            continue

        # Validate date format
        date_obj = pd.to_datetime(row['date'])

        # Extract category (standardize)
        category = row['category'].strip()

        # Confidence score
        confidence = 0.95
        if pd.isna(row['po_number']):
            confidence -= 0.02
        if row['approval_status'] != 'Approved':
            confidence -= 0.05

        extracted.append({
            'transaction_id': row['transaction_id'],
            'date': row['date'],
            'merchant_original': row['merchant'],
            'merchant_cleaned': merchant_clean,
            'category': category,
            'amount': amount,
            'status': row['status'],
            'department': row['department'],
            'extraction_confidence': confidence,
            'data_quality': 'Good' if confidence > 0.9 else 'Fair'
        })
    except Exception as e:
        errors.append({'id': row['transaction_id'], 'error': str(e)})

# Save extracted data
extracted_df = pd.DataFrame(extracted)
extracted_df.to_csv('extracted_5000.csv', index=False)

print(f"✅ Extracted: {len(extracted):,} records")
print(f"❌ Errors: {len(errors)}")
print(f"Success rate: {(len(extracted)/len(df)*100):.2f}%")
print(f"\nConfidence distribution:")
print(extracted_df['extraction_confidence'].describe())
print(f"\nFirst 10 extracted:")
print(extracted_df.head(10))
```
